```python
p0 = 1/13
p1 = 1.5/13
n = 3000

p0_likely = True
arr = np.linspace(p0,p1,100)
i = 0
while p0_likely:
    if (stats.binom.pmf(round(n*arr[i]), n, p0) < stats.binom.pmf(round(n*arr[i]), n, p1)):
        p0_likely = False
    else:
        i += 1
p = arr[i]

print(p0,p1,p)
print(stats.binomtest(round(p*n), n, p0, alternative='greater'))
print(stats.binomtest(round(p*n), n, p1, alternative='less'))
```
